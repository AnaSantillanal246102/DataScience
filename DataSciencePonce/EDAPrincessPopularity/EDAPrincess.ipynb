```python
#Make a copy of the categorical variables
categoricals_var=df.select_dtypes(include=['object'])
categoricals_var.head()
```

##Analysis and Visualization

###Numerical Variables

```python
#This line show the unique values of the column of Princess Name
df['PrincessName'].unique()
```

First, I ask for the name of the princess that we have on the dataset, because we need have a preview of the posibles results

```python
df['PrincessName'].value_counts()
```

Mulan is the princess with more rows in this dataset, however we could thinking that is a fact of Popularity but with this analysis dont have a high score of Popularity.

```python
#Here we show the popularity score and the princess's name in a descending order
df[['PrincessName','PopularityScore']].sort_values(by='PopularityScore', ascending=False).head(10)
```

We can notice that many have the same score, so we will calculate the popularity using other facts, such as the Number of MerchItems on Amazon, the BoxOffice data, and other numerical variables, that will be meaning, it give us a clear number of their results in the daata set, so we group the values by the Princess Name coincidence.

```python
#We define the columns that won't be used
DontUse=['FirstMovieYear','MovieRuntimeMinutes','NumberOfSongs']
#First, drop the columns in the array
#calculamos el promedio por columna y agrupamos segun el nombre de la princesa
Popularity=numericals_var.drop(columns=DontUse).groupby('PrincessName').mean()
Popularity.head()
```

```python
#We create a cycle that repeat the process of create and show the plot
#This cycle is repeated the number ofr columns before the last one
for column in Popularity.columns[:-1]:
  #create the plot of bar with the column and the Precess's Name
    fig = px.bar(Popularity, x=column, y="PrincessName", color="PrincessName")
  #show the plot
    fig.show()
```

These graphs show a discrepancy, because the previous plot shows that Princess Elsa have the highest mean of score popularity and before when we present the highers scores dont have a enough score.

```python
#create arrays for save the max values
Popularity.reset_index(inplace=True)
#create the arrays
columns= []
princess= []
values= []
#create a cycle that will repeated the same times as columns
for column in Popularity.columns:
  #Ask for skip th column 'PrincessName'
    if column != 'PrincessName':
      #Ask the maximum value
        maxValue= Popularity[column].max()
        #save the princess's name of the max value in this column
        maxPrincess= Popularity[Popularity[column] == maxValue]['PrincessName'].values
        #add to the arrays
        columns.append(column)
        princess.append(maxPrincess)
        values.append(maxValue)
#create a dataframe with the arrays
MaxPopularity= pd.DataFrame({'Column': columns, 'PrincessName': princess, 'Value': values})
#set the column 'Column' as the index
MaxPopularity.set_index('Column', inplace=True)
MaxPopularity
```

We can see that several princesses have the maximum in various factors, to determine which princess is the most popular, we decided to average all the columns for have a more specific score of popularity.

```python
#We create a new column that save the mean of all the row (axis=1)
Popularity.drop(columns='PrincessName', inplace=True)
Popularity['MeanScoresPopularity']=Popularity.mean(axis=1)
Popularity.head()
```

```python
#First we sort the values
Popularity_plot=Popularity.sort_values(by='MeanScoresPopularity').reset_index()
#we create the plot and give as variables the Name of the princess and the mean of popularity
fig = px.bar( Popularity_plot,x='PrincessName',y='MeanScoresPopularity',
             color='PrincessName',title='Average of Popularity by Princess')
fig.show()
```

```python
#Calculate the standard deviation
std_por_columna = Popularity.std()
std_por_columna
```

At first sight there is no great separation between the scores, but when we calculate the standar deviation between the columns is clear that columns as NumMerchItemsOnAmazon (Number of MerchItems in Amazon), BoxOfficeMillions (MIllion of BoxOffice) are the columns where we have more difference between Princesses.

```python
#Create the plot for standard deviation
fig = px.bar(std_por_columna, x=std_por_columna.index, y=std_por_columna.values,
             color=std_por_columna.index)
fig.show()
```

All the columns have a relative influence in the Popularity of Princess.

```python
#We normalized the dataframe for after make a heatmap and not have a big difference in numbers
PopularityNormal=(Popularity-Popularity.min())/(Popularity.max()-Popularity.min())
#then sort by score of popularity
PopularityNormal=PopularityNormal.sort_values(by='MeanScoresPopularity')
PopularityNormal.head()
```

```python
#We create a heatmap for a best visualitation of the Popularity Socre
fig = px.imshow(PopularityNormal*100, labels=dict(x="Facts", y="Princess Name", color="Popularity %"), aspect="auto")
fig.show()
```

In this heatmap we can see more clearly the columns that give to each princess her score and why Tiana have the most highest score because she have a higher score in RottenTomatoScore and NumMerchItemsOnAmazon.

The relationship between popularity score and some characteristics like whether a princess has a solo song could give us insight into how these categorical variables influence a character's popularity.

```python
#we add the score popularity to the categorical's variables because we want found a relation in both variables
categoricals_var['ScorePopularity']=df['PopularityScore']
#first we found the maximum score and save it
maxscore=categoricals_var['ScorePopularity'].max()
#then we search all the results with the same score
bymaxScore=categoricals_var[categoricals_var['ScorePopularity']==maxscore]
bymaxScore
```

Here we have the rows with the highest score, so we will compare with variables like HasSoloSong, IsRoyalByBirth, HasMagicalPowers, IsBasedOnRealStory, SpeaksToAnimals, MarriedByEnd;
We think that those are characteristics that could give us interesting information.

```python
#Define the columns that we will analysis
CategoricalBinary=bymaxScore[['HasSoloSong', 'IsRoyalByBirth',
                                   'HasMagicalPowers','IsBasedOnRealStory',
                                   'SpeaksToAnimals','MarriedByEnd']]
#create a figure with 6 subplots (2*3)
fig= make_subplots(rows=2, cols=3, subplot_titles=CategoricalBinary.columns)
#Make a cycle for repeat the number of columns
for i,column in enumerate(CategoricalBinary[:6]):
  #This asign a number of row and columns
  row = (i // 3) + 1
  col = (i % 3) + 1
  #We count the number of times that a value was repeated in the maximum results previous
  count = bymaxScore[column].value_counts().reset_index()
  #Asign to the dataframe the values
  count.columns = [column, 'Count']
  #Create a plot with the data of the dataframe
  bar = go.Bar(x=count[column], y=count['Count'], name=column)
  #Asign to the figure the plot with they respective rows and columns
  fig.add_trace(bar, row=row, col=col)
fig.show()
```

* In the IsBasedOnRealStory and SpeaksToAnimals plots, we don't have information because there is a coincidence (the same number of princesses) in both classifications.

* In the HasMagicalPowers plot we have a bias to the Princesses that have Powers.
*  In the HasSoloSong, IsRoyalByBirt and MarriedByEnd plots we have a negative inclination, so we can't take this as a fact of influence.

In the same way, we want to compare with the minimum score, so we prepare the sames columns and the ScorePopularity.

```python
#Ask for the minimum score
minscore=categoricals_var['ScorePopularity'].min()
#Save the rows with the minimum score
byminScore=categoricals_var[categoricals_var['ScorePopularity']==minscore]
byminScore
```

```python
#Define the columns that we will analyze
CategoricalBinary=byminScore[['HasSoloSong', 'IsRoyalByBirth',
                                   'HasMagicalPowers','IsBasedOnRealStory',
                                   'SpeaksToAnimals','MarriedByEnd']]
#Create a figure where we will show the 6 subplots (2*3)
fig= make_subplots(rows=2, cols=3, subplot_titles=CategoricalBinary.columns)
#Create a cycle for the number of columns
for i,column in enumerate(CategoricalBinary[:6]):
  #This asign a number of row and columns
  row = (i // 3) + 1
  col = (i % 3) + 1
  #here, count the number of times that is repeated th values
  count = byminScore[column].value_counts().reset_index()
  #Add to a dataframe the values
  count.columns = [column, 'Count']
  #create a plot with the dataframe values
  bar = go.Bar(x=count[column], y=count['Count'], name=column)
  #Asign to the figure the plot with they repsective rows and columns
  fig.add_trace(bar, row=row, col=col)
fig.show()
```

In this case:
* The HasSoloSong and IsBasedOnRealStory plots say that this aspects are not a positive factor for popularity
* IsRoyalByBirth, HasMagicalPowers, SpeaksToAnimals and MarriedByEnd  plots don´t give us specific information because have the same influence.
