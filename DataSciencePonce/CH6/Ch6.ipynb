###**csv.Dialect**
To define a new format with a different delimiter, string quoting convention, or line terminator, we could define a simple subclass of csv.Dialect

```python
class my_dialect(csv.Dialect):
    lineterminator = "\n"
    delimiter = ";"
    quotechar = '"'
    quoting = csv.QUOTE_MINIMAL
```

```python
f= open("examples/ex7.csv")
r=csv.reader(f)
for line in r:
  print(line)
```

```python
f= open("examples/ex7.csv")
r1=csv.reader(f, dialect=my_dialect)
for line in r1:
  print(line)
```

```python
f= open("examples/ex7.csv")
f1=f.read()
f2=f1.replace(',','|')
r2= csv.reader(f2, delimiter="|")
for line in r2:
  print(line)
```

##**XML** (Extensible Markup Language)

###**Parsing XML with lxml.objectify**
The xml file have a type of structure data, where each data have a determinate sort that use python for transform data to dataframes or series, using tags for assign the order

##**lxml.objectify (librarie)**

###**objectify.parse()**
This transform the data to data readable by python

```python
from lxml import objectify
path = "examples/Performance_MNR.xml"
with open(path) as f:
  parsed = objectify.parse(f)
```

### **.getroot()**
this save the main root of the file

```python
root = parsed.getroot()
root
```

```python
root.INDICATOR
```

```python
data = []
#here assign in a list the name of columns that will be ignore
skip_fields = ["PARENT_SEQ", "INDICATOR_SEQ",
               "DESIRED_CHANGE", "DECIMAL_PLACES"]
#este ciclo recorre las secciones en la rama principal
for elt in root.INDICATOR:
  #define un diccionario para agregar datos luego
    el_data = {}
    #este ciclo recorre los elementos hijo de la rama principal
    for child in elt.getchildren():
      #compara la etiqueta de las columnas con las que se iban a ignorar
        if child.tag in skip_fields:
          #en casso de encontrarse coincidencia sale del bucle
            continue
            #caso contrario asigna al diccionario los valores
        el_data[child.tag] = child.pyval
        #finalmente agrega el diccionario a la lista data
    data.append(el_data)

el_data
```

```python
pf=pd.DataFrame(data)
pf.head()
```

```python
data[0]
```

#Using HDF5 Format
HDF5 is a respected file format intended for storing large quantities of scientific array data.
* It is available as a C library.
* The “HDF” in HDF5 stands for hierarchical data format.
* Each HDF5 file can store multiple datasets and supporting metadata.

HDF5 can be a good choice for working with datasets that don't fit into memory, as you can efficiently read and write small sections of much larger arrays.

**You must first install PyTables by installing the tables package**

###**pandas.HDFStore()**

```python
import numpy as np
frame = pd.DataFrame({"a": np.random.standard_normal(100)})
store = pd.HDFStore("examples/mydata.h5")
store["obj1"] = frame
store["obj1_col"] = frame["a"]
store
```

```python
store["obj1"]
```

HDFStore recibe 2 parametros, "table" and "fixed", "table" is slower but this allow query operations

##**.put()**
The put is an explicit version of the store["obj2"] = frame method but allows us to set other options like the storage format

```python
store.put("obj2", frame, format="table")
store.select("obj2", where=["index >= 10 and index <= 15"])
store.close()
```

###**pandas.read_hdf()**

##**sqlite3**(librarie)

```python
import sqlite3
#this is a sentence of SQL
query = """CREATE TABLE test1 (a VARCHAR(20), b VARCHAR(20),
c REAL, d INTEGER);"""
#This generate a conection with the database
con = sqlite3.connect("mydata.sqlite")
#this execute the sentece of SQL that create a table
con.execute(query)
```

**.commit()** this is lika a checkpoint that save the changes permanently

```python
#just can create once table
con.commit()
```

```python
data = [("Atlanta", "Georgia", 1.25, 6),
        ("Tallahassee", "Florida", 2.6, 3),
        ("Sacramento", "California", 1.7, 5)]
stmt = "INSERT INTO test VALUES(?, ?, ?, ?)"
con.executemany(stmt, data)
```

```python
con.commit()
```

**.cursor** is an object that was crated when you execute a query

**.fetchall** Is a function that take all the rows of the query

```python
cursor = con.execute("SELECT * FROM test")
# a cursor is an object that is created when execute a query
#fetchall take all the results
rows = cursor.fetchall()
rows
```
