#Adjusting the spacing around subplots

subplots_adjust(left=None, bottom=None, right=None, top=None, wspace=None, hspace=None)

**wspace hspace** (percent)

```python
fig, axes = plt.subplots(2, 2, sharex=True, sharey=True)
for i in range(2):
    for j in range(2):
        axes[i, j].hist(np.random.standard_normal(200), bins=10,
                        color="black", alpha=0.5)
fig.subplots_adjust(wspace=0, hspace=0)
```

## sharex & sharey

```python
fig, axes = plt.subplots(2, 2, sharex=True, sharey=False)
for i in range(2):
    for j in range(2):
        axes[i, j].hist(np.random.standard_normal(200), bins=10,
                        color="red", alpha=0.2)
fig.subplots_adjust(wspace=.2, hspace=.2)
```

```python
import pandas as pd
from datetime import datetime

fig, ax = plt.subplots()

data = pd.read_csv("examples/spx.csv", index_col=0, parse_dates=True)
spx = data["SPX"]

spx.plot(ax=ax, color="black")

crisis_data = [
    (datetime(2007, 10, 11), "Peak of bull market"),
    (datetime(2008, 3, 12), "Bear Stearns Fails"),
    (datetime(2008, 9, 15), "Lehman Bankruptcy")
]

for date, label in crisis_data:
    ax.annotate(label, xy=(date, spx.asof(date) + 75),
                xytext=(date, spx.asof(date) + 225),
                arrowprops=dict(facecolor="red", headwidth=5, width=3,
                                headlength=5),
                horizontalalignment="left", verticalalignment="top")

# Zoom in on 2007-2010
ax.set_xlim(["1/1/2007", "1/1/2011"])
ax.set_ylim([600, 1800])

ax.set_title("Important dates in the 2008–2009 financial crisis")
```

###.asof()
This method returns the last available value before or on the given date.
This is useful when you don't have data exactly on the requested date, but you want to use the closest previous value (very common in financial data, where data isn't available every day—for example, on weekends or holidays).

```python
spx[datetime(2008, 3, 12)]
```

## **Patches**

```python
fig, ax = plt.subplots()

rect = plt.Rectangle((0.1, 0.7), 0.6, 0.2, color="black", alpha=0.3)
circ = plt.Circle((0.7, 0.3), 0.20, color="blue", alpha=0.3)
pgon = plt.Polygon([[0.15, 0.15], [0.35, 0.4], [0.2, 0.6]],
                   color="green", alpha=0.5)

ax.add_patch(rect)
ax.add_patch(circ)
ax.add_patch(pgon)
```

```python
fig, ax = plt.subplots()

pgon = plt.Polygon([[0.15, 0.15], [0.2, 0.2], [0.3, 0.3], [.4,.5],[.8,.6]],
                   color="green", alpha=0.5)
ax.add_patch(pgon)
```

##pandas.crosstab()

```python
tips = pd.read_csv("examples/tips.csv")
tips.head()
```

```python
party_counts = pd.crosstab(tips["day"], tips["size"])
party_counts = party_counts.reindex(index=["Thur", "Fri", "Sat", "Sun"])
party_counts
```

```python
party_counts = party_counts.loc[:, 2:5]
party_counts
```

##.div()

```python
# Normalize to sum to 1
#party_counts.sum(axis="columns") calcula la suma total de cada fila
#.div(..., axis="index") divide cada fila del DataFrame original por su suma correspondiente (por día).
party_pcts = party_counts.div(party_counts.sum(axis="columns"),axis="index")
party_pcts
```

```python
party_pcts.plot.bar(stacked=True)
```

##*Facet Grids and Categorical Data

One way to visualize data with many categorical variables is to use a facet grid, which is a two-dimensional layout of plots where the data is split across the plots on each axis based on the distinct values of a certain variable.
###.catplot()

categorical plot
simplifies making many kinds of faceted plots split by categorical variable

```python
sns.catplot(x="day", y="tip_pct", hue="time", col="smoker",
            kind="bar", data=tips[tips.tip_pct < 1])
```

```python
sns.catplot(x="day", y="tip_pct", row="time",col="smoker",
            kind="bar", data=tips[tips.tip_pct < 1])
```

##box plot
box plots (which show the median, quartiles, and outliers) can be an effective visualization type.

```python
sns.catplot(x="tip_pct", y="day", kind="box",
            data=tips[tips.tip_pct < 0.5])
```
