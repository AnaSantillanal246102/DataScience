**Puntaje Crediticio**

El puntaje crediticio (credit score) es un número que resume la capacidad de una persona para cumplir con sus obligaciones financieras. Se calcula a partir del historial crediticio de la persona e influye directamente al momento de querer adquirir productos financieros como tarjetas de crédito o préstamos.

Es relevante principalmente para la industria financiera pues les permite calcular riesgos de pérdidas y categorizar clientes.

    En USA el credit score se encuentra entre los 300 y 850 puntos done 850 es la más alta confiabilidad de pago.

**Objetivo del análisis**

Observar cuales sonn los principales factores que afectan en el puntaje crediticio, encontrar similitudes o diferencias entre los clientes que nos ayuden a formar una postura entre las diferentes capacidades financieras comparadas con su puntaje crediticio.

```python
#creamos una variables en la cual evaluaremos segun el cliente por tanto creamos una copia ordenada de las columnas numericas
bycustomer=num_var.sort_values(by='credit_score', ascending=False)
#agregamos el id del cliente que nos va a permitir agrupar los datos
bycustomer['customer_id']=df['customer_id']
#agrupamos segun el cliente promediando las coincidencias
bycustomer=bycustomer.groupby('customer_id').mean()
#convertimos eel indice a columna
bycustomer.reset_index( inplace=True)
#creamos una figura en la cual agregaremos un grafico que relacione la edad con el historial crediticiio
fig = px.scatter(bycustomer,x="age", y="credit_history_age",color="credit_history_age",color_continuous_scale="Cividis",
                 title="Relación entre Edad y el Historial Crediticio",hover_data=["credit_score","customer_id"],trendline="ols")
#cambiamos el tamaño de los ciculitos (marcas)
fig.update_traces(marker=dict(size=5))
#guardamos el id del outlier en una variables
outlierid = "CUS_0x7aba"
#buscamos el registro con ese id
outlierdata = bycustomer[bycustomer['customer_id'] == outlierid].iloc[0]
#agregamos anotaciones para señalar el outlier con unaa flecha y que indique el ID
fig.add_annotation(x=outlierdata['age'],y=outlierdata['credit_history_age'],text=f"Outlier ID: {outlierid}",arrowhead=2,
    arrowsize=1,arrowwidth=1.5,arrowcolor="black")
fig.show()
```

###Edad

```python
#creamos un dataframe que cree una copia de las columnas numericas ordenada por puntaje crediticio
byage=num_var.sort_values(by='credit_score', ascending=False)
#agrupamos segun la edad
byage=byage.groupby('age').mean()
byage.head()
```

```python
#declaramos en una lista las columnas que queremos analizar
cols = ['annual_income','num_credit_card','interest_rate','num_of_loan','num_of_delayed_payment',
        'num_credit_inquiries','outstanding_debt','monthly_balance','credit_score']
byage.reset_index(inplace=True)
#creamos una figura como cuadricula 3*3 donde agreguemos cada grafica y cambiaremos los _ en el nombre de las columnas
fig = make_subplots(rows=3, cols=3,subplot_titles=[col.replace("_", " ").title() for col in cols])
#inicializamos las columnas y filas en 1
row = 1
col = 1
#creamos un ciclo que repita el grafico el numero de columnas
for i, column in enumerate(cols):
  #agregamos el grafico a la figura
  fig.add_trace(
      #asignamos el grafico la columna y la edad
      go.Scatter(x=byage['age'], y=byage[column],mode='markers',marker=dict(color=byage['age'],
                  colorscale='Turbo'),name=column),row=row, col=col)
  #sumamos uno a la columna despues de cada grafica
  col += 1
  #cuando tengamos las 3 columnas se asignara otravez el valor a 1 pero se le agregara una fila
  if col > 3:
      col = 1
      row += 1
#cambiamos las dimensiones de presentacion
fig.update_layout(showlegend=False)
fig.show()
```

####Conclusiones de los gráficos por edad:
* **Ingreso Anual** (annual_income). Podemos ver que hay una estrecha relación entre la edad y los ingresos anuales, encontrando que entre mayor edad mayor es el ingreso.

* **Numero de tarjetas de credito** (num_credit_card). Contrario a los ingresos entre más jovenes sean los clientes tienen una mayor cantidad de tarjetas de credito podemos considerar que al contar con menores ingresos requieren de instrumentos crediticios como tarjetaas para manejar sus ingresos o a la intención de crear un historial crediticio.
* **Tasa de intéreses** (interest_rate). Tambien notamos que la tasa de intéreses se reduce con la edad, que puede tener relación a sus ingresos o la madurez de manejo de herramientas crediticias, por tanto los bancos tienen mayor confianza en estos clientes.
* **Número de prestamos** (num_of_loan). Tenemos un comportamiento parecido a la cantidad de tarjetas de crédito, por lo que apoya nuestra teoría de que como la edad afecta a su uso.
* **Número de pagos pendientes** (num_of_delayed_payment). Reafirmando lo anterior las personas con mayor edaad suelen cumplir con sus pagos de una forma considerablemente mayor a las personas jovenes.
* **Numero de solicitudes de crédito** (num_credit_inquiries). Aqui mantenemos un mayor movimiento entre las personas jovenes, por tanto podemos considerar esta variable como datos que refuerzan al número de préstamos, tambien podemos considerar una posible brecha de edad en el uso de herramientas digitales para consulta de crédito..
* **Deuda Pendiente** (outstanding_debt). Para reafirmar la variable de número de pagos pendientes, las personas más jovenes tienen mayor dificultad para pagar, que puede deberse a su bajo ingreso.
* **Balance mensual** (monthly_balance). Las personas de edad avanzada, cuentan con números más altos en su balance mensual, puede ser a razón de sus ingresos mensuales, y consideraando que son los que menos préstamos tienen, lo que lo hace congruente.
* **Puntaje Crediticio** (credit_score). Finalmente, el propósito de nuestro estudio se ve reflejado en esta grafica por edad, obtenemos que entre mayor edad el puntaje crediticio tambien aumenta, algunas de las varibales pueden ser los ingresos, los tipos de gastos, la confiaabilidad con la banca, el historial crediticio.

Tambien observamos que hay dos puntos críticos, las personas que llegan a los 20 años comienzan a tener un comportamiento similar en todas las graficas, y las personas despúes de los 45 años tienen aparentemente una mayor estabilidad economica.

###Ocupacion

```python
#creamos una copia que de las columnas numericas ordenadas por el puntaje crediticio
byoccupation=num_var.sort_values(by='credit_score', ascending=False)
#agregamos la comuna de ocupacion
byoccupation['occupation']=cat_var['occupation']
#agrupamos por ocupacion
byoccupation=byoccupation.groupby('occupation').mean()
byoccupation.head()
```

```python
#determinamos cuales son las columnass que vamos a analizar
cols = ['annual_income','interest_rate','num_of_loan','num_credit_inquiries','outstanding_debt','credit_score']
#creamos una figura con cuadricula 2*3 donde agreguemos cada grafica y cambiaremos los _ en el nombre de las columnas
fig = make_subplots(rows=2, cols=3,subplot_titles=[col.replace("_", " ").title() for col in cols])
#inicializamos las columnas y filas a 1
row = 1
col = 1
#creamos un ciclo que repita el proceso el numero de columnass
for i, column in enumerate(cols):
  #agregamos el grafico a la figura
    fig.add_trace(
        #creamos el grafico de barras con la ocupacion y la columna
        go.Bar(x=byoccupation['occupation'], y=byoccupation[column],name=column,
               ),row=row, col=col)
    #agregamos 1 a la columna despues de cada grafico creado
    col += 1
    #cuando las columnas sean mayor a 2 se vuelven a acomodar a 1 pero se cambia de fila
    if col > 3:
        col = 1
        row += 1
#cambiamos las dimensiones de la figura
fig.update_layout(height=800,width=1300,showlegend=False)
fig.show()
```

####Conclusiones de los gráficos por Ocupación:
* **Ingreso Anual** (annual_income). Podemos destacar que los clientes con ocupación de arquitectos son los que tienen mayor ingresos anuales y los Periodistas son quienes perciben menores ingresos anuales.
* **Tasa de intéreses** (interest_rate). Tambien notamos que la tasa de intéreses más baja en relación a la anterior afirmación es de los arquitectos y la más alta es para los abogados, esto nos crea una condicionante pues apesar de que los abogados no se encuentra como ocupacion con ingresos mas bajos, puede ser a razón de incumplimientos de pagos.
* **Número de prestamos** (num_of_loan). Nuevamente los arquitectos son la ocupación que menos préstamos pide y los doctores son quienes más cantidad de préstamos tienen, lo que nos crea una condicionante de en que gastan los médicos.
* **Numero de solicitudes de crédito** (num_credit_inquiries). Aquí presentamos un cambio de profesión siendo los músicos y mecanicos, quienes tienen menos y más solicitudes de crédito (respectivamente).
* **Deuda Pendiente** (outstanding_debt). Vuelve a protagonizar la ocupación de aarquitecto con menor deuda y mecanico con mayor deuda a una diferencia de casi $300.
* **Puntaje Crediticio** (credit_score). Finalmente, los Media Managers son la ocupación con mayor puntaje crediticio y en el otro extremo los Escritores, que puede deberse a la naturaleza de este trabajo, pues hay muchos picos en la carrera de un escritor.

###Mes

```python
#creamos una copia de las columnas numericas
bymonth=num_var.sort_values(by='credit_score', ascending=False)
#agregamos la columna de mes
bymonth['month']=cat_var['month']
#agrupamos segun el mes
bymonth=bymonth.groupby('month').mean()
bymonth
```

```python
#designamos las columnas que queremos analizar
columnasmes=['total_emi_per_month','num_credit_inquiries']
bymonth.reset_index(inplace=True)
#creamos una figura donde agregaremos los graaficos en una cuadricula de 1*2
fig = make_subplots(rows=1, cols=2,subplot_titles=[col.replace("_", " ").title() for col in columnasmes])
#Inicializamos las columnas a 1
row = 1
col = 1
#creamos un ciclo que repita el proceso el numero de columnas
for i, column in enumerate(columnasmes):
  #agregamos a la figura el grafico
    fig.add_trace(
        #creamos un grafico de barras con el mes y la columna
        go.Bar(y=bymonth['month'], x=bymonth[column],name=column,orientation="h"),row=row, col=col)
    #agregamos a 1 la columna
    col += 1
#cambiamos la dimension de figura
fig.update_layout(showlegend=False)
fig.show()
```

* **EMI por mes**

        EMI(Equated Monthly Installment). Es una tarifa dispuesta por el banco que esta en relación al préstamo o crédito y que tiene que ser saldada mensualmente.

Cuando analizamos los registros mensualmente podemos obtener que Agosto es el mes con mayor EMI, que puede asociarse a variables sociodemográficas puesto que coincide con el regreso escolar,  ademas de que se encuentra a mitad de año a difrencia de Enero que es el mes con menor EMI, que al igual puede asociarse a variaables como prestaciones y aguinaldos.
* **Número de solicitudes de crédito**

Podemos notar que en relación con el análisis anterior el mes con mayor solicitudes es Agosto y Enero el mes con menor solicitudes.

```python
#creamos un dataframe para guardar el typo de comportamiento
bypayment=pd.DataFrame()
#agregamos el comportamiento de pago
bypayment['payment_behaviour']=cat_var['payment_behaviour']
#Agregamos el puntaje crediticio
bypayment['credit_score']=num_var['credit_score']
#agrupamos segun el comportamiento
bypayment = bypayment.groupby('payment_behaviour').mean()
#ordenamos segun el puntaje crediticio
bypayment = bypayment.sort_values(by='credit_score', ascending=False)
bypayment
```

###Comportamiento

```python
bypayment.reset_index(inplace=True)
fig = px.bar(bypayment, x='payment_behaviour', y='credit_score',
             color='credit_score')
fig.show()
```

El comportamiento de pago es de las variables más importantes para el puntaje crediticio, lo podemos ver claro en esta gráfica, además nos otorga dos características clave para formar un buen puntaje crediticio.

Podemos notar que entre más gastos mayor es el puntaje y que aunque se tengan pagos pequeños tiene mayor puntaje que aquellos que gastan poco y pagan cantidades altas.
Esto nos declara que los puntajes crediticios más altos son de clientes que no solo pagan si no que tambien gastan bastante.

Con lo anterior en mente, nos gustaría cuestionar a que tipo de préstamos se les atribuye el mayor puntaje crediticio.

```python
typeloan = pd.DataFrame()
typeloan['type_of_loan'] = cat_var['type_of_loan'].str.strip("[]").str.split(", ")
typeloan['credit_score'] = num_var['credit_score']
 #Descompone la lista en filas
typeloan = typeloan.explode('type_of_loan')
# Elimina espacios
typeloan['type_of_loan'] = typeloan['type_of_loan'].str.strip()
typeloan['loan_counts'] = typeloan['type_of_loan'].map(typeloan['type_of_loan'].value_counts())
# Agrupa por tipo de préstamo y calcula promedio
typeloan = typeloan.groupby('type_of_loan').mean()
typeloan
```

```python
typeloan.sort_values(by='credit_score', ascending=False, inplace=True)
fig = px.bar(typeloan, x='type_of_loan', y='credit_score',
             color='type_of_loan',
             title='Puntaje Crediticio Promedio por Tipo de Préstamo')
fig.show()
```

A pesar de que no existe una gran diferencia notamos que aquellos registros que no tenían préstamos o no eran especificados eran los que contaban con mayor puntaje crediticio, sin embargo no podemos tener una conclusión concreta de esto puesto que estas eran las respuestas con mayor repetición.
