Ejemplo 02: Valores atípicos y estimados de variabilidad

```python
# Importar librerías
import pandas as pd

# Cargar el dataset
df = pd.read_csv('/content/drive/MyDrive/DataScienceBedu/Datasets/S01/Ejemplo_01_Titanic_Dataset.csv') # Modifica la ruta de acuerdo a tu entorno de trabajo

# Mostrar las primeras filas del dataset
df.head()
```

```python
# Calcular la varianza de la tarifa
varianza_fare = df['Fare'].var()
print(f'La varianza de la tarifa de los pasajeros en el Titanic es: {varianza_fare:,.2f}')

# Nota: Este valor representa el valor de tarifa elevado al cuadrado, por lo que no tiene unidades monetarias.
```

```python
# Calcular la desviación estándar de la tarifa
desviacion_estandar_fare = df['Fare'].std()
print(f'La desviación estándar de la tarifa de los pasajeros en el Titanic es: ${desviacion_estandar_fare:,.2f}')
```

```python
# Calcular el rango de la edad
edad_max = df['Age'].max()
edad_min = df['Age'].min()
rango_edad = edad_max - edad_min
print(f'El rango de la edad de los pasajeros en el Titanic es: {rango_edad}')
```

```python
from scipy import stats
import pandas as pd

# Eliminar valores NaN en la columna 'Age'
edad_sin_nan = df['Age'].dropna()

# Calcular la media truncada
media_truncada_edad = stats.trim_mean(edad_sin_nan, 0.05)
print(f'La media truncada de la edad de los pasajeros en el Titanic es: {media_truncada_edad:.2f}')
```

Ejemplo 03: Cuartiles y Percentiles

```python
# Importar librerías
import pandas as pd

# Cargar el dataset
df = pd.read_csv('/content/drive/MyDrive/DataScienceBedu/Datasets/S01/Ejemplo_01_Titanic_Dataset.csv') # Modifica la ruta de acuerdo a tu entorno de trabajo

# Calcular los cuartiles de la tarifa
Q1_fare = df['Fare'].quantile(0.25)
Q2_fare = df['Fare'].quantile(0.50)
Q3_fare = df['Fare'].quantile(0.75)

print(f'Primer cuartil (Q1) de la tarifa: ${Q1_fare:.2f}')
print(f'Segundo cuartil (Q2) de la tarifa (mediana): ${Q2_fare:.2f}')
print(f'Tercer cuartil (Q3) de la tarifa: ${Q3_fare:.2f}')
```

```python
# Eliminar valores NaN en la columna 'Age'
edad_sin_nan = df['Age'].dropna()

# Calcular el percentil 90 de la edad
percentil_90_edad = edad_sin_nan.quantile(0.90)

print(f'El percentil 90 de la edad de los pasajeros en el Titanic es: {percentil_90_edad:.2f} años')
```

```python
df = pd.read_csv('/content/drive/MyDrive/DataScienceBedu/Datasets/S01/Reto_01_Laptops_Dataset.csv')
# Calcular la media
promedio_precio = df['price'].mean()
print(f'El promedio del precio de las laptops fue: ${promedio_precio:,.2f}')
# Calcular la mediana
mediana_precio = df['price'].median()
print(f'El valor central del precio de las laptops es: ${mediana_precio:,.2f}')
# Calcular la moda
moda_precio = df['price'].mode()[0]
print(f'El precio más común de las laptops es: ${moda_precio:,.2f}')
```

Este analísis nos indica que la media se inclina a la derecha, sumado a que la moda predomina hacia la izquierda del valor central, nos indica que hay outliers muy distantes de la media, porque la moda se encuentra a la izquierda entonces sabemos que no se tratan de muchos datos que afecten del lado derecho.

```python
# Calcular la media
promedio_size_screen = df['screen_size'].mean()
print(f'El promedio del tamaño de las pantallas de las laptops fue: {promedio_size_screen}')
# Calcular la mediana
mediana_size_screen = df['screen_size'].median()
print(f'El valor central del tamaño de las pantallas de las laptops es: {mediana_size_screen}')
# Calcular la moda
moda_size_screen = df['screen_size'].mode()[0]
print(f'El tamaño de las pantallas más común de las laptops es: {moda_size_screen}')
```

En cambio este analisis, nos indica que el promedio no es distante del valor central y es reforzado porque la moda compagina con el valor central.

```python
df = pd.read_csv('/content/drive/MyDrive/DataScienceBedu/Datasets/S01/Reto_02_Salary_Professions.csv')
varianza_salary = df['SALARY'].var()
print(f'La varianza de los salarios entre profesiones es: ${varianza_salary:,.2f}')
desviacion_estandar_salary = df['SALARY'].std()
print(f'La desviación estándar de los salarios entre profesiones es de: ${desviacion_estandar_salary:,.2f}')
salary_max = df['SALARY'].max()
salary_min = df['SALARY'].min()
rango_salary = salary_max - salary_min
print(f'El rango de los salarios de las profesiones es: ${rango_salary:,.2f}')
```

La varianza de los datos es alta, lo que nos indica que hay una dispercion entre los datos, adicional la desviacion estandar nos indica que las diferencias entre los datos y la media es de ~36,000, finalmente el rango nos demuestra que la distancia entre salarios es bastante amplia, lo que apoya a las medidas de dispersion.

```python
# Calcular la media truncada
media_salary = df['SALARY'].mean()
print(f'La media de los salarios es: ${media_salary:.2f}')
```

```python
from scipy import stats
import pandas as pd
salary_sin_nan = df['SALARY'].dropna()
# Calcular la media truncada
media_truncada_salary = stats.trim_mean(salary_sin_nan, 0.05)
print(f'La media truncada de los salarios es: ${media_truncada_salary:.2f}')
```

Hay una diferencia de ~$6,000, lo que es un valor considerable por lo que podemos deducir que existen algunos outliers afectando la media original.

✅ Desafío adicional: Calcula el rango intercuartílico (IQR) y determina si existen valores atípicos en la columna StudyTimeWeekly.

```python
df = pd.read_csv('/content/drive/MyDrive/DataScienceBedu/Datasets/S01/Reto_03_Student_Performance.csv')
# Calcular los cuartiles de la tarifa
Q1_StudyTimeWeekly = df['StudyTimeWeekly'].quantile(0.25)
Q2_StudyTimeWeekly = df['StudyTimeWeekly'].quantile(0.50)
Q3_StudyTimeWeekly = df['StudyTimeWeekly'].quantile(0.75)
print(f'Primer cuartil (Q1) del tiempo de estudio semanal: {Q1_StudyTimeWeekly:.2f} horas')
print(f'Segundo cuartil (Q2) del tiempo de estudio semanal (mediana): {Q2_StudyTimeWeekly:.2f} horas')
print(f'Tercer cuartil (Q3) del tiempo de estudio semanal: {Q3_StudyTimeWeekly:.2f} horas')
# Eliminar valores NaN en la columna 'Age'
StudyTimeWeekly_sin_nan = df['StudyTimeWeekly'].dropna()
percentil_25_StudyTimeWeekly = StudyTimeWeekly_sin_nan.quantile(0.25)
percentil_40_StudyTimeWeekly = StudyTimeWeekly_sin_nan.quantile(0.40)
percentil_90_StudyTimeWeekly = StudyTimeWeekly_sin_nan.quantile(0.90)
print(f'El percentil 25 del tiempo de estudio semanal es: {percentil_25_StudyTimeWeekly:.2f} horas')
print(f'El percentil 40 del tiempo de estudio semanal es: {percentil_40_StudyTimeWeekly:.2f} horas')
print(f'El percentil 90 del tiempo de estudio semanal es: {percentil_90_StudyTimeWeekly:.2f} horas')
```

La mayoría de los estudiantes estudian entre 5 y 14 horas semanales. La mediana de 9.71 h indica que el tiempo de estudio típico está cerca de las 10 horas por semana. Solo un pequeño grupo (10%) estudia más de 17.79 horas, lo cual puede representar a los estudiantes más comprometidos, exigidos o con más tiempo disponible.

```python
IQR = Q3_StudyTimeWeekly - Q1_StudyTimeWeekly

limite_inferior = Q1_StudyTimeWeekly - 1.5 * IQR
limite_superior = Q3_StudyTimeWeekly + 1.5 * IQR

# Filtrar outliers altos
outliers = df[df['StudyTimeWeekly'] > limite_superior]

print(f'IQR: {IQR:.2f} horas')
print(f'Límite superior: {limite_superior:.2f} horas')
print(f'Número de valores atípicos altos: {len(outliers)}')
```
