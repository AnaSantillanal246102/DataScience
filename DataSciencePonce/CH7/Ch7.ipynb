## Function or Mapping

```python
data = pd.DataFrame({"food": ["bacon", "pulled pork", "bacon",
                     "pastrami", "corned beef", "bacon",
                     "pastrami", "honey ham", "nova lox"],
                     "ounces": [4, 3, 12, 6, 7.5, 8, 3, 5, 6]})
data
```

```python
#you define the type of meat in a dictionary
meat_to_animal = {
  "bacon": "pig",
  "pulled pork": "pig",
  "pastrami": "cow",
  "corned beef": "cow",
  "honey ham": "pig",
  "nova lox": "salmon"
}
# guardas en la columna 'animal' la agrupacion de la funcion map
#la funcion map asocia con la comida food con el diccionario.
#usa 'food' como clave al buscar en el diccionario
#guarda solo el valor definido en el diccionario
data["animal"] = data["food"].map(meat_to_animal)
data
```

```python
#this function do the same thing asthe previous one
def get_animal(x):
  return meat_to_animal[x]
data["food"].map(get_animal)
```

**Note:**
In the string representation of an interval, a parenthesis means that the side is open (exclusive), while the square bracket means it is closed (inclusive).
You can change which side is closed by passing right=False

```python
pd.cut(ages, bins, right=False)
```

##Computing Indicator/Dummy Variables
Key Words: dummy or indicator matrix

**Dummy Variables**. It is a binary (0 or 1) variable created to represent categorical data as numerical data.

**Indicator matrix**.  Typically refers to a matrix of binary values (0 or 1), where each row corresponds to an observation, and each column corresponds to a specific category or feature. It "indicates" the presence or absence of a particular characteristic.

If a column in a DataFrame has k distinct values, you would derive a matrix or DataFrame with k columns containing all 1s and 0s. pandas has a pandas.get_dummies function for doing this

```python
df = pd.DataFrame({"key": ["b", "b", "a", "c", "a", "b"],
                    "data1": range(6)})
df
```

###pd.get_dummies

```python
pd.get_dummies(df["key"], dtype=float)
```

```python
pd.get_dummies(df["key"])
```

```python
pd.get_dummies(df["key"], dtype=int)
```

####.join()
this add a prefix

```python
dummies = pd.get_dummies(df["key"], prefix="key", dtype=float)
df_with_dummy = df[["data1"]].join(dummies)
df_with_dummy
```
